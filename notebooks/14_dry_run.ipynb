# 📘 Agentic Architectures 14: Observability + Dry-Run Harness

Welcome to a critical notebook in our series, focused on deployment and operational safety for AI agents. We'll implement an **Observability and Dry-Run Harness** — an important pattern for testing, debugging, and safely governing agents that interact with real-world systems.

The core principle is simple but powerful: **never run an agent's action against a live environment without knowing exactly what it's about to do.** This architecture formalizes a "look before you leap" process. The agent first executes its plan in `dry_run` mode, which doesn't change the real world but generates detailed logs and a clear action plan. That plan is then presented to a human (or an automated checker) for approval before the final, live execution is allowed.

To demonstrate this, we'll build a **Corporate Social Media Agent**. This agent is tasked with drafting and publishing posts. We'll see how the dry-run harness lets us:
1.  **Generate a Proposed Post:** The AI creatively drafts a post based on a prompt.
2.  **Perform a Dry Run:** The agent calls the `publish` function in safe `dry_run=True` mode, generating logs of what *would* happen.
3.  **Human-in-the-Loop Review:** A human operator is shown the exact post content and the dry-run trace. They must type `approve` to proceed.
4.  **Execute the Live Action:** Only after approval is the `publish` function called again, this time with `dry_run=False`, to perform the real action.

This pattern is the foundation of responsible agent deployment, providing the transparency and control needed to safely operate AI in production.

### Definition
An **Observability and Dry-Run Harness** is a testing and deployment architecture that intercepts an agent's actions. It first executes them in a "dry run" or "sandbox" mode that simulates the action without triggering real consequences. The resulting plan and logs are then presented for review, and only after explicit approval is the action executed against the live environment.

### High-level workflow

1.  **Agent proposes an action:** The agent determines a plan or a specific tool call to execute (e.g. `api.post_update(...)`).
2.  **Dry-run execution:** The harness calls the agent's plan with a `dry_run=True` flag. The underlying tools are designed to recognize this flag and only report what they *would have* done, along with logs and traces.
3.  **Collecting observability data:** The harness captures the proposed action, the dry-run logs, and any other relevant trace data from the simulation.
4.  **Human/automated review:** This observability data is presented to a reviewer. A human can check for correctness, safety, and alignment with goals. An automated system can check for policy violations or known bad patterns.
5.  **Go/No-Go decision:** The reviewer decides to `approve` or `reject`.
6.  **Live execution (on 'Go'):** If approved, the harness re-executes the agent's action, this time with `dry_run=False`, letting it take effect in the real world.

### When to use it / Applications
*   **Debugging and testing:** During development, to understand exactly how an agent interprets a task and what actions it takes, without side effects.
*   **Production validation and safety:** As a standing production feature for any agent that can change state, spend money, send messages, or perform any other irreversible action.
*   **CI/CD for agents:** Integrating the dry-run harness into an automated testing pipeline to validate agent behavior before deploying new versions.

### Strengths and weaknesses
*   **Strengths:**
    *   **Maximum transparency and safety:** Provides a clear, auditable preview of an agent's actions, preventing costly or embarrassing mistakes.
    *   **Excellent for debugging:** Makes it easy to trace an agent's logic and tool calls without leaving real-world changes to undo.
*   **Weaknesses:**
    *   **Delays deployment/execution:** The mandatory review step (especially a human one) introduces latency, making it unsuitable for real-time applications.
    *   **Requires tool support:** The tools and APIs the agent uses must be designed to support a `dry_run` mode.

## Phase 1: Building the environment and tools

The core of this architecture is a tool that supports a `dry_run` mode. We'll build a simple `SocialMediaAPI` class. Its `publish_post` method will behave differently depending on the `dry_run` flag, giving us the observability we need.

```python
console = Console()

# A structured model for the agent's proposed post
class SocialMediaPost(BaseModel):
    content: str = Field(description="The full text of the social media post.")
    hashtags: List[str] = Field(description="A list of relevant hashtags, without the '#' symbol.")

# Key component: a tool with a dry_run flag
class SocialMediaAPI:
    """A mock social media API that supports a dry-run mode."""

    def publish_post(self, post: SocialMediaPost, dry_run: bool = True) -> Dict[str, Any]:
        """Publishes a post to the social media feed."""
        timestamp = datetime.datetime.now().isoformat()
        hashtags_str = ' '.join([f'#{h}' for h in post.hashtags])
        full_post_text = f"{post.content}\n\n{hashtags_str}"

        if dry_run:
            # In dry-run mode we don't execute, we just return the plan and logs
            log_message = f"[DRY RUN] At {timestamp}, would publish the following post:\n--- PREVIEW ---\n{full_post_text}\n--- END PREVIEW ---"
            console.print(Panel(log_message, title="[yellow]Dry Run Log[/yellow]", border_style="yellow"))
            return {"status": "DRY_RUN_SUCCESS", "log": log_message, "proposed_post": full_post_text}
        else:
            # In live mode we perform the action
            log_message = f"[LIVE] At {timestamp}, successfully published post!"
            console.print(Panel(log_message, title="[green]Live Execution Log[/green]", border_style="green"))
            # A real API call would go here, e.g. twitter_client.create_tweet(...)
            return {"status": "LIVE_SUCCESS", "log": log_message, "post_id": f"post_{hash(full_post_text)}"}

social_media_tool = SocialMediaAPI()
print("SocialMediaAPI tool with dry-run support defined successfully.")
```

## Phase 2: Building the Dry-Run Harness with LangGraph

Now we'll build the full workflow. The graph will manage the process state, moving from proposing an action to a dry-run-and-review step, and finally to conditional execution based on human approval.

```python
llm = ChatNebius(model="mistralai/Mixtral-8x22B-Instruct-v0.1", temperature=0.5)

# LangGraph state
class AgentState(TypedDict):
    user_request: str
    proposed_post: Optional[SocialMediaPost]
    dry_run_log: Optional[str]
    review_decision: Optional[str] # 'approve' or 'reject'
    final_status: str

# Graph nodes
def propose_post_node(state: AgentState) -> Dict[str, Any]:
    """The creative agent that drafts a social media post."""
    console.print("--- 📝 SMM Agent drafting a post ---")
    prompt = ChatPromptTemplate.from_template(
        "You are a creative and engaging social media manager for a major AI company. Based on the user's request, draft a compelling social media post, including relevant hashtags.\n\nRequest: {request}"
    )
    post_generator_llm = llm.with_structured_output(SocialMediaPost)
    chain = prompt | post_generator_llm
    proposed_post = chain.invoke({"request": state['user_request']})
    return {"proposed_post": proposed_post}

def dry_run_review_node(state: AgentState) -> Dict[str, Any]:
    """Performs the dry run and requests human review."""
    console.print("--- 🧐 Performing Dry Run and awaiting human review ---")
    dry_run_result = social_media_tool.publish_post(state['proposed_post'], dry_run=True)

    # Present the plan for review
    review_panel = Panel(
        f"[bold]Proposed Post:[/bold]\n{dry_run_result['proposed_post']}",
        title="[bold yellow]Human-in-the-Loop: Review Required[/bold yellow]",
        border_style="yellow"
    )
    console.print(review_panel)

    # Get human approval
    decision = ""
    while decision.lower() not in ["approve", "reject"]:
        decision = console.input("Type 'approve' to publish, or 'reject' to cancel: ")

    return {"dry_run_log": dry_run_result['log'], "review_decision": decision.lower()}

def execute_live_post_node(state: AgentState) -> Dict[str, Any]:
    """Executes the live publish after approval."""
    console.print("--- ✅ Post approved, executing live ---")
    live_result = social_media_tool.publish_post(state['proposed_post'], dry_run=False)
    return {"final_status": f"Post successfully published! ID: {live_result.get('post_id')}"}

def post_rejected_node(state: AgentState) -> Dict[str, Any]:
    """Handles the case where a post is rejected."""
    console.print("--- ❌ Post rejected by reviewer ---")
    return {"final_status": "The action was rejected by the reviewer and was not executed."}

# Conditional edge
def route_after_review(state: AgentState) -> str:
    """Routes to execution or rejection based on the human review."""
    return "execute_live" if state["review_decision"] == "approve" else "reject"

# Build the graph
workflow = StateGraph(AgentState)
workflow.add_node("propose_post", propose_post_node)
workflow.add_node("dry_run_review", dry_run_review_node)
workflow.add_node("execute_live", execute_live_post_node)
workflow.add_node("reject", post_rejected_node)

workflow.set_entry_point("propose_post")
workflow.add_edge("propose_post", "dry_run_review")
workflow.add_conditional_edges("dry_run_review", route_after_review, {"execute_live": "execute_live", "reject": "reject"})
workflow.add_edge("execute_live", END)
workflow.add_edge("reject", END)

dry_run_agent = workflow.compile()
print("Dry-Run Harness agent graph compiled successfully.")
```

## Phase 3: Demonstration

Let's test the full system. First with a safe, standard request that we'll approve. Second, with a more ambiguous request that might generate a risky post, which we'll reject.

```python
def run_agent_with_harness(request: str):
    initial_state = {"user_request": request}
    # Note: you'll be prompted to type in the console below this cell.
    result = dry_run_agent.invoke(initial_state)
    console.print(f"\n[bold]Final Status:[/bold] {result['final_status']}")

# Test 1: A safe post we'll approve.
console.print("--- ✅ Test 1: Safe post (Approve) ---")
run_agent_with_harness("Draft a positive launch announcement for our new AI model, 'Nebula'.")

# Test 2: A risky post we'll reject.
console.print("\n--- ❌ Test 2: Risky post (Reject) ---")
run_agent_with_harness("Draft a post that emphasizes how much better our new 'Nebula' model is than the competition.")
```

### Analyzing the results

The demonstration is a perfect example of the harness's value:

1.  **The safe post:** The first request was straightforward. The agent generated a professional, enthusiastic post. The dry run showed a preview of exactly what would be published. We approved it, and the `[LIVE]` log confirms the real action was taken. The process worked as intended.

2.  **The risky post:** The second request was more ambiguous and could be interpreted aggressively. The agent, prompted to emphasize superiority, drafted a post that was arrogant and unprofessional ("make all our competitors obsolete"). While the agent followed its creative prompt, this isn't a message a real company would want to publish.

This is where the harness proved its value. The dry run surfaced this risky content *before* it could be published. The human reviewer easily spotted the inappropriate tone and typed `reject`. The graph correctly routed to `post_rejected_node`, and the final status confirms that **no live action was taken.** A potential PR crisis was averted by a simple, structured workflow.

This cleanly separates the agent's creative, unpredictable generation from deterministic, controlled execution, providing a vital safety layer.

## Conclusion

In this notebook we built a complete **Observability and Dry-Run Harness**. This architecture isn't just a feature — it's a foundational philosophy for deploying agents that interact with the real world. By applying a `propose -> review -> execute` loop, we gain critical benefits:

- **Transparency:** We know exactly what the agent intends to do before it does it.
- **Control:** We have a "human in the loop" (or an automated policy engine) with final veto power over every action.

While this pattern introduces latency, the safety and reliability it provides are non-negotiable for most real-world applications. It's an essential tool for any developer building reliable, trustworthy, production-ready agentic systems.
