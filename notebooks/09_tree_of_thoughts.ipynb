# 📘 Agentic Architectures 9: Tree-of-Thoughts Planning

Welcome to the ninth notebook in our series. Today we explore a powerful reasoning and planning architecture: **Tree-of-Thoughts (ToT)**. This pattern elevates an agent's problem-solving ability from a linear chain of thought to a multi-path, exploratory search.

Instead of generating a single, sequential line of reasoning, a ToT agent generates several candidate "thoughts," or next steps, at each stage of a problem. It then evaluates those thoughts, pruning invalid or unpromising branches and expanding the most promising ones. This builds a search tree where the agent can backtrack, explore alternatives, and systematically navigate a complex problem space.

To demonstrate this, we'll give our agent a classic logic puzzle: the **Wolf, Goat, and Cabbage problem**. This puzzle is famous because it requires non-obvious steps (e.g. bringing an item *back*) and has several invalid states that can trap a naive solver. We'll show how a simple **Chain-of-Thought** agent can fail, while a **ToT** agent methodically builds a valid plan by exploring and scoring the tree of possibilities.

### Definition
**Tree-of-Thoughts (ToT)** is an agentic reasoning framework where problem-solving is modeled as a tree search. The agent explores several reasoning paths (branches) at once. At each step it generates candidate next steps ("thoughts"), scores their viability, and decides which paths to keep exploring, effectively pruning the search space.

### High-level workflow

1.  **Decomposition:** The problem is broken into a series of steps or thoughts.
2.  **Thought Generation:** For the current problem state, the agent generates several candidate next steps or thoughts. This creates branches in the search tree.
3.  **State Evaluation:** Every new thought (leading to a new state) is scored by a "critic" or validation function. This scoring can check:
    *   **Validity:** Is this move allowed by the problem's rules?
    *   **Progress:** Does this move bring us closer to a solution?
    *   **Heuristic:** How likely is this path to succeed?
4.  **Pruning & Expansion:** Invalid or unpromising branches are pruned. The agent then continues with the most promising active branches, repeating the thought-generation process.
5.  **Solution:** The process continues until a goal state is reached. The solution is the path of thoughts from the root to the goal.

### When to use it / Applications
*   **Logic puzzles and math problems:** Problems with clear rules and goal states that require multi-step, non-linear reasoning (e.g. Sudoku, river-crossing puzzles).
*   **Complex planning:** When a task requires a detailed plan where the order of operations matters and constraints must be respected (e.g. planning a complex trip with multiple legs and budget constraints).
*   **Creative writing or code generation:** Exploring several plot branches or implementation strategies before settling on one.

### Strengths and weaknesses
*   **Strengths:**
    *   **Reliability:** Systematically explores the problem space, making it far less likely to get stuck or return a wrong answer compared to a single-pass method.
    *   **Handles combinatorial complexity:** Well suited to problems where the number of possible sequences is huge.
*   **Weaknesses:**
    *   **Computationally heavy:** Requires significantly more LLM calls and state management than a simple Chain-of-Thought prompt, making it slower and more expensive.
    *   **Needs a good evaluator:** Search effectiveness depends heavily on the quality of the state-evaluation logic.

## Phase 1: Defining the problem environment

A Tree-of-Thoughts system needs a well-defined environment to operate in. For our "Wolf, Goat, and Cabbage" puzzle, this means we need to programmatically define:

1.  **State representation:** A way to describe the location of every item.
2.  **Validation rules:** A function to check whether a state is invalid (e.g. the goat and cabbage left alone together).
3.  **Goal state:** A way to check whether the puzzle is solved.
4.  **Possible moves:** A function to determine every legal move from a given state.

```python
console = Console()

class PuzzleState(BaseModel):
    "Represents the state of the 'Wolf, Goat, and Cabbage' puzzle."
    # Use sets for the unordered collections of items on each bank
    left_bank: set[str] = Field(default_factory=lambda: {"wolf", "goat", "cabbage"})
    right_bank: set[str] = Field(default_factory=set)
    boat_location: str = "left"
    move_description: str = "Initial state."

    def is_valid(self) -> bool:
        """Checks whether the current state is valid (nothing has been eaten)."""
        # Check the left bank
        if self.boat_location == "right":
            if "wolf" in self.left_bank and "goat" in self.left_bank:
                return False
            if "goat" in self.left_bank and "cabbage" in self.left_bank:
                return False
        # Check the right bank
        if self.boat_location == "left":
            if "wolf" in self.right_bank and "goat" in self.right_bank:
                return False
            if "goat" in self.right_bank and "cabbage" in self.right_bank:
                return False
        return True

    def is_goal(self) -> bool:
        """Checks whether the goal state has been reached."""
        return self.right_bank == {"wolf", "goat", "cabbage"}

    def __hash__(self):
        # Make the state hashable so we can check for visited states
        return hash((frozenset(self.left_bank), frozenset(self.right_bank), self.boat_location))

    def __eq__(self, other):
        return self.__hash__() == other.__hash__()

def get_possible_moves(state: PuzzleState) -> list[PuzzleState]:
    """Generates every possible valid next state from the current one."""
    moves = []
    current_bank = state.left_bank if state.boat_location == "left" else state.right_bank

    # Option 1: Carry one item in the boat
    for item in current_bank:
        # Optimization: Create new state directly instead of deep copy + mutate
        if state.boat_location == "left":
            new_state = PuzzleState(
                left_bank=state.left_bank - {item},
                right_bank=state.right_bank | {item},
                boat_location="right",
                move_description=f"Move {item} to the right bank."
            )
        else:
            new_state = PuzzleState(
                left_bank=state.left_bank | {item},
                right_bank=state.right_bank - {item},
                boat_location="left",
                move_description=f"Move {item} to the left bank."
            )
        if new_state.is_valid():
            moves.append(new_state)

    # Option 2: Move the boat empty
    # Optimization: Create new state directly
    if state.boat_location == "left":
        empty_move_state = PuzzleState(
            left_bank=state.left_bank,
            right_bank=state.right_bank,
            boat_location="right",
            move_description="Move the boat empty to the right bank."
        )
    else:
        empty_move_state = PuzzleState(
            left_bank=state.left_bank,
            right_bank=state.right_bank,
            boat_location="left",
            move_description="Move the boat empty to the left bank."
        )
    if empty_move_state.is_valid():
        moves.append(empty_move_state)

    return moves

print("Puzzle environment defined successfully.")
```

## Phase 2: Implementing the ToT agent with LangGraph

Now we'll build the agent itself. Our graph's state will track every active path (branch) in our tree of thoughts. The nodes will perform ToT's key actions:

1.  **Expand paths (Thought Generator):** An LLM-backed node that looks at each active path's latest state and proposes a promising next move from the list of valid possibilities.
2.  **Prune paths (State Evaluator):** This node tidies up after generation. It removes any paths that entered an invalid state or a cycle (revisiting a previous state).
3.  **Check for a solution (Goal Check):** A conditional node that checks whether any active path has reached the goal state. If so, it ends the loop.

```python
llm = ChatNebius(model="mistralai/Mixtral-8x22B-Instruct-v0.1", temperature=0.4)

# Pydantic model for the LLM's move choice
class MoveChoice(BaseModel):
    best_move_index: int = Field(description="Index of the best move from the provided list of possible moves.")
    reasoning: str = Field(description="A brief justification for why this move is the most promising.")

# LangGraph state
class ToTState(TypedDict):
    problem_description: str
    # Each path is a list of PuzzleState objects
    active_paths: List[List[PuzzleState]]
    # We'll store the final solution here
    solution: Optional[List[PuzzleState]]

# Graph nodes
def initialize_search(state: ToTState) -> Dict[str, Any]:
    """Node for setting up the initial search state."""
    initial_puzzle_state = PuzzleState()
    return {"active_paths": [[initial_puzzle_state]]}

def expand_paths(state: ToTState) -> Dict[str, Any]:
    """The 'Thought Generator'. Expands every active path with a promising next move."""
    console.print("--- Expanding Paths ---")
    new_paths = []
    choice_llm = llm.with_structured_output(MoveChoice)

    prompt = ChatPromptTemplate.from_messages([
        ("system", "You are an expert logic puzzle solver. Your goal is to solve the Wolf, Goat, and Cabbage problem. Analyze the current path and choose the most promising next move from the list of options to reach the goal."),
        ("human", "Problem: {problem}\n\nCurrent Path History:\n{path_history}\n\nFrom the final state, choose the best next move from this list:\n{possible_moves}")
    ])

    for path in state['active_paths']:
        last_state = path[-1]
        possible_next_states = get_possible_moves(last_state)

        if not possible_next_states:
            continue # This path is a dead end

        path_history_str = " -> ".join([s.move_description for s in path])
        possible_moves_str = "\n".join([f"{i}: {s.move_description}" for i, s in enumerate(possible_next_states)])

        # For simplicity, and to demonstrate breadth, we explore several moves here.
        # A more advanced ToT could use the LLM to pick just the single best one.
        # Here we let every valid move branch out, to demonstrate the tree structure.
        for next_state in possible_next_states:
            new_paths.append(path + [next_state])

    console.print(f"[cyan]Expanded to {len(new_paths)} candidate paths.[/cyan]")
    return {"active_paths": new_paths}

def prune_paths(state: ToTState) -> Dict[str, Any]:
    """The 'State Evaluator'. Prunes paths that are invalid or contain a cycle."""
    console.print("--- Pruning Paths ---")
    pruned_paths = []
    for path in state['active_paths']:
        # Check for cycles: if the last state has already appeared earlier in the path
        if path[-1] in path[:-1]:
            continue # A cycle was found, prune this path

        # get_possible_moves already guarantees validity, but this is a good place for extra checks.
        pruned_paths.append(path)

    console.print(f"[green]Pruned down to {len(pruned_paths)} valid, cycle-free paths.[/green]")
    return {"active_paths": pruned_paths}

# Conditional node
def check_for_solution(state: ToTState) -> str:
    """Checks whether any path has reached the goal, and routes execution."""
    for path in state['active_paths']:
        if path[-1].is_goal():
            console.print("[bold green]Solution found![/bold green]")
            # Side effect: update the solution in the state. LangGraph carries this out.
            state['solution'] = path
            return "solution_found"
    return "continue_search"

# Build the graph
workflow = StateGraph(ToTState)

workflow.add_node("initialize", initialize_search)
workflow.add_node("expand", expand_paths)
workflow.add_node("prune", prune_paths)

workflow.set_entry_point("initialize")
workflow.add_edge("initialize", "expand")
workflow.add_edge("expand", "prune")

workflow.add_conditional_edges(
    "prune",
    check_for_solution,
    {
        "solution_found": END,
        "continue_search": "expand"
    }
)

tot_agent = workflow.compile()
print("Tree-of-Thoughts agent graph compiled successfully.")
```

## Phase 3: Demonstration and analysis

Now let's run our ToT agent on the puzzle. We'll compare its systematic approach against a simple, single-pass Chain-of-Thought query to highlight the difference in reliability.

```python
problem = "A farmer wants to cross a river with a wolf, a goat, and a cabbage. The boat can only carry the farmer and one other item. The farmer cannot leave the wolf alone with the goat, nor the goat alone with the cabbage. How can the farmer get everyone across safely?"

console.print("--- 🌳 Running the Tree-of-Thoughts agent ---")
# The recursion limit prevents infinite loops in our graph
config = {"recursion_limit": 15}
final_state = tot_agent.invoke({"problem_description": problem}, config=config)

console.print("\n--- ✅ ToT Agent's Solution ---")
if final_state.get('solution'):
    solution_path = final_state['solution']
    # Use rich.Tree for a nice visual output
    tree = Tree("[bold magenta]Solution path for the Wolf, Goat, and Cabbage puzzle[/bold magenta]")
    for i, state in enumerate(solution_path):
        tree.add(f"[green]{i+1}.[/green] {state.move_description}")
    console.print(tree)
else:
    console.print("[bold red]No solution found within the step limit.[/bold red]")

console.print("\n--- 🤔 Running the simple Chain-of-Thought agent ---")
cot_prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a world-class logic puzzle solver. Provide a step-by-step solution to the user's puzzle."),
    ("human", "{problem}")
])
cot_chain = cot_prompt | llm
cot_result = cot_chain.invoke({"problem": problem}).content
console.print(Markdown(cot_result))
```

### Analyzing the results

The difference between the two approaches is profound:

- **Chain-of-Thought (CoT):** This approach relies on the LLM's pretrained knowledge to recall the solution. For a classic, widely known problem like this, a capable LLM can often get the right answer on the first try. But if it makes even one mistake, it has no mechanism for self-correction. For a novel or more complex problem, the odds of failure are far higher. Its correctness is a matter of memory, not verifiable reasoning.

- **Tree-of-Thoughts (ToT):** This agent *discovered* the solution through a systematic, verifiable search. It didn't just recall the answer; it built it. We can see the process in the logs: expanding paths, then pruning the ones that hit a dead end or a cycle. Even if the LLM driving expansion made a suboptimal choice on one branch, the agent could keep exploring other, more promising branches. This method is far more reliable and trustworthy, because its final solution is guaranteed valid according to the rules of the environment we defined.

The ToT agent's success rests not on luck or memorization, but on the soundness of its search algorithm. That makes it a significantly more robust approach for tasks that demand high reliability and planning.
